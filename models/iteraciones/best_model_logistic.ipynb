```python
# MEJOR MODELO DEL MOMENTO
# Mejor metrica C=6     | Train: 0.9913 | Val: 0.2710   ->  base

import re
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression


def cleant_text_no_simbols(texto):
    if not isinstance(texto, str):
        return ""

    texto = texto.lower()
    texto = re.sub(r'[-¬]\s*', '', texto)
    texto = re.sub(r'[^a-z0-9áéíóúüñ¡!¿?.,;():«»—+ſꝛũ’\s]', ' ', texto)    
    texto = re.sub(r'\s+', ' ', texto).strip()

    return texto


# Lectura archivos
data = pd.read_csv("../../data/train.csv")
data_test = pd.read_csv("../../data/eval.csv")


col_text_clean = "text_clean"

# Limpieza columna text
data[col_text_clean] = data['text'].apply(cleant_text_no_simbols)
data_test[col_text_clean] = data_test['text'].apply(cleant_text_no_simbols)

# Separación de la matrix X y la columna y
data_X = data[col_text_clean]
data_y = data["decade"]

# Creación del set de datos de train y validation
X_train, X_val, y_train, y_val = train_test_split(
    data_X, data_y, test_size=0.2, random_state=42, stratify=data_y
)

# Construcción del pipeline
valores_C = [1, 5, 6, 7, 8, 9, 10, 15]
historia_train = []
historia_val = []

# ¡Aquí empieza la indentación clave!
for c in valores_C:
    modelo_prueba = make_pipeline(
        TfidfVectorizer(
            analyzer='char_wb',
            ngram_range=(2, 4),
            max_df=0.85,
            min_df=2,            
            max_features=150000, 
            lowercase=False,
            sublinear_tf=True
        ),
        
        LogisticRegression(
            penalty='l2',
            solver='lbfgs',
            C=c,
            class_weight='balanced',
            random_state=42,
            max_iter=5000
        )
    )
    
    modelo_prueba.fit(X_train, y_train)
    acc_train = accuracy_score(y_train, modelo_prueba.predict(X_train))
    acc_val = accuracy_score(y_val, modelo_prueba.predict(X_val))
    
    historia_train.append(acc_train)
    historia_val.append(acc_val)
    
    print(f"C={c:<5} | Train: {acc_train:.4f} | Val: {acc_val:.4f}")
```

```python
MEJOR_C = 6

modelo_texto = make_pipeline(
    TfidfVectorizer(
        analyzer='char_wb',
        ngram_range=(2, 4),
        max_df=0.85,
        min_df=2,            
        max_features=150000, 
        lowercase=False,
        sublinear_tf=True
    ),

    LogisticRegression(
        penalty='l2',
        solver='lbfgs',
        C=MEJOR_C,         
        class_weight='balanced',
        random_state=42,
        max_iter=5000
    )
)

modelo_texto.fit(X_train, y_train)

accuracy_train = accuracy_score(y_train, modelo_texto.predict(X_train))
accuracy_val = accuracy_score(y_val, modelo_texto.predict(X_val))

print(f"\n=== DIAGNÓSTICO DEL MODELO FINAL (C={MEJOR_C}) ===")
print(f"Accuracy Train: {accuracy_train:.4f}")
print(f"Accuracy Val: {accuracy_val:.4f}")

# --- 4. RETRENAMIENTO Y EXPORTACIÓN KAGGLE ---
print("\nRetrenando con el 100% de los datos para el envío final...")
modelo_texto.fit(data_X, data_y)

predictions_test = modelo_texto.predict(data_test[col_text_clean])

df_envio = pd.DataFrame({
    'id': data_test['id'],
    'answer': predictions_test
})

df_envio.to_csv(f"model_linear_{accuracy_val:.4f}.csv", index=False)
print("¡Archivo answer_logit.csv generado con éxito!")
```
